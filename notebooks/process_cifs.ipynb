## Finding useful keys:

- `_chemical_melting_point` - exact melting point
- `_chemical_melting_point_gt` - if exact not known, this is value, that it is greater then
- `_chemical_melting_point_lt` - same, but lower then

```python
cifs_path = "/Users/aleksandr.varlamov/cif/all_cifs"
cifs = sorted(os.listdir(cifs_path))
cifs_id = list(map(lambda x: x.split(".")[0], cifs))
```

```python
# Create a list to store the data
melting_points_data = []

for i, id in enumerate(tqdm(cifs_id, total=len(cifs_id))):
    path = os.path.join(cifs_path, f"{id}.cif")
    try:
        cif = cif_from_file(path)
        # Check if melting point exists in the cif dictionary
        melting_point = cif.get('_chemical_melting_point', None)
        # Replace '?' with None
        if melting_point == '?':
            melting_point = None
        melting_points_data.append({'id': id, 'T': melting_point})
    except Exception as e:
        print(f"Error reading {path}: {e}")
        melting_points_data.append({'id': id, 'T': None})
        continue

# Create a DataFrame and save to CSV
melting_points_df = pd.DataFrame(melting_points_data)
melting_points_df.to_csv('melting_points.csv', index=False)
```
